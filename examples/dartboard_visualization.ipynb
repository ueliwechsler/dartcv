# 2D Dartboard Visualization

This notebook renders a clean synthetic dartboard template that can be used as the basis for a computer vision pipeline.


```python
from pathlib import Path
import matplotlib.pyplot as plt
from dartcv.board import render_dartboard, render_svg
```

```python
fig, ax = render_dartboard(figsize=(6,6), show=False)
plt.show()
```

```python
svg = render_svg(width=800, height=800)
svg_path = Path('dartboard_template.svg')
svg_path.write_text(svg, encoding='utf-8')

print(f'Saved SVG to: {svg_path.resolve()}')
```

## Notes

- The generated board is geometrically regular and therefore useful as a template for CV calibration and alignment.
- In a pipeline, you would detect the board in a camera image, estimate its center and radius, and use a perspective transform before scoring darts.
- The next step is usually to match a live board image to this synthetic template.


## Compact rendering batch

This feature generates boards, rather than consuming photographs. Select
rendering sizes below and set `RUN_BATCH = True`; only final renders appear.
For photo-set inspection use `color_dart_pipeline_overview.ipynb`.

```python
import matplotlib.pyplot as plt
from dartcv.board import render_dartboard

RUN_BATCH = False
BATCH_FIGURE_SIZES = [(4, 4), (6, 6), (8, 8)]
if RUN_BATCH:
    for figure_size in BATCH_FIGURE_SIZES:
        batch_figure, batch_axis = render_dartboard(figsize=figure_size, show=False)
        batch_axis.set_title(f"Final board render {figure_size}")
        display(batch_figure)
        plt.close(batch_figure)
```
